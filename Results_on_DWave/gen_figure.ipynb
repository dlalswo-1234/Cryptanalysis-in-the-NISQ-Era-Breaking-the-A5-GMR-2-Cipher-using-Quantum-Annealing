```python
import os
import re
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1.inset_locator import inset_axes
def extract_data_from_file(file_path):
    try:
        with open(file_path, 'r', encoding='utf-8') as file:
            content = file.read()
            
        # Extract qpu_access_time
        qpu_pattern = r'qpu_access_time:\s+([\d.]+)\s*ms'
        qpu_match = re.search(qpu_pattern, content)
        
        # Extract attack success status
        attack_pattern = r'Attack Success:\s+(True|False)'
        attack_match = re.search(attack_pattern, content)
        
        result = {}
        result['qpu_access_time'] = float(qpu_match.group(1)) if qpu_match else None
        result['attack_success'] = attack_match.group(1) == 'True' if attack_match else None
        
        return result
            
    except Exception as e:
        print(f"Error reading {file_path}: {e}")
        return None

def main():

    base_paths = [
   "./_Experiment_Result_KS1/run_data"
    ]
    
    # Collect all data
    all_qpu_times = []
    all_attack_success = 0
    total_files = 0
    
    for base_path in base_paths:
        print(f"Processing: {os.path.basename(base_path)}")
        
        for i in range(1, 101):
            test_folder = f"test_{i:03d}"
            file_path = os.path.join(base_path, test_folder, "results.txt")
            
            if os.path.exists(file_path):
                data = extract_data_from_file(file_path)
                if data:
                    total_files += 1
                    
                    if data['qpu_access_time'] is not None:
                        all_qpu_times.append(data['qpu_access_time'])
                    
                    if data['attack_success'] is True:
                        all_attack_success += 1
            else:
                print(f"File not found: {file_path}")
    
    # Calculate statistics
    if all_qpu_times:
        mean_time = np.mean(all_qpu_times)
        std_time = np.std(all_qpu_times, ddof=1)
        
        print(f"\n" + "="*50)
        print("COMBINED RESULTS")
        print("="*50)
        print(f"Total files processed: {total_files}")
        print(f"QPU times collected: {len(all_qpu_times)}")
        print(f"Attack successes: {all_attack_success}")
        print(f"Attack success rate: {all_attack_success/total_files:.2%}")
        print(f"\nQPU Access Time Statistics:")
        print(f"  Mean: {mean_time:.6f} ms")
        print(f"  Standard Deviation: {std_time:.6f} ms")
        print(f"  Min: {min(all_qpu_times):.6f} ms")
        print(f"  Max: {max(all_qpu_times):.6f} ms")
        print(f"  Median: {np.median(all_qpu_times):.6f} ms")
        
        # Create plots
        plt.figure(figsize=(15, 10))
    
        # Time series
        plt.subplot(2, 3, 3)

        n = len(all_qpu_times)
        x = np.arange(1, n + 1)

        plt.scatter(x, all_qpu_times, color='tab:blue', s=6,alpha=0.8,
                 linewidth=1.5, label='QPU Access Time')

        plt.fill_between(x, mean_time - std_time, mean_time + std_time,
                         alpha=0.2, color='orange')

        plt.axhline(mean_time, color='red', linestyle='--', linewidth=1.5,
                    alpha=0.8, label='Mean')

        y_min = max(0, mean_time - 1.5 * std_time)
        y_max = mean_time + 1.5 * std_time
        plt.ylim(y_min, y_max)
        plt.xlim(1, n)

        min_idx = int(np.argmin(all_qpu_times))
        min_x = min_idx + 1
        min_y = all_qpu_times[min_idx]

        marker_y = y_min + 0.5
        plt.plot(min_x, marker_y, marker=r'$\downarrow$', color='purple', markersize=12)

        plt.text(1, marker_y + 1, 
         rf"Lowest: {min_y:.2f} ms (\#{min_x})",   # ← \# 사용
         fontsize=9, color='purple', ha='left', va='bottom')


        plt.xlabel('Test Number')
        plt.ylabel('QPU Access Time (ms)')
        plt.legend(loc='upper right')
        plt.grid(True, alpha=0.3)
        plt.tight_layout()
        plt.savefig('qpu_time_1.png', format='png', dpi=1200, bbox_inches='tight')



if __name__ == "__main__":
    main()
```

```python
import os
import re
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.offsetbox import AnchoredText

def extract_data_from_file(file_path):
    try:
        with open(file_path, 'r', encoding='utf-8') as file:
            content = file.read()

        qpu_pattern = r'qpu_access_time:\s+([\d.]+)\s*ms'
        qpu_match = re.search(qpu_pattern, content)

        attack_pattern = r'Attack Success:\s+(True|False)'
        attack_match = re.search(attack_pattern, content)

        result = {}
        result['qpu_access_time'] = float(qpu_match.group(1)) if qpu_match else None
        result['attack_success'] = attack_match.group(1) == 'True' if attack_match else None

        return result
    except Exception as e:
        print(f"Error reading {file_path}: {e}")
        return None

def main():
    base_paths = [
   "./_Experiment_Result_KS3/run_data"
    ]

    all_qpu_times = []
    all_attack_success = 0
    total_files = 0

    for base_path in base_paths:
        print(f"Processing: {os.path.basename(base_path)}")
        for i in range(1, 101):
            test_folder = f"test_{i:03d}"
            file_path = os.path.join(base_path, test_folder, "results.txt")

            if os.path.exists(file_path):
                data = extract_data_from_file(file_path)
                if data:
                    total_files += 1
                    if data['qpu_access_time'] is not None:
                        all_qpu_times.append(data['qpu_access_time'])
                    if data['attack_success']:
                        all_attack_success += 1
            else:
                print(f"File not found: {file_path}")

    if all_qpu_times:
        mean_time = np.mean(all_qpu_times)
        std_time = np.std(all_qpu_times, ddof=1)

        print(f"\n" + "=" * 50)
        print("COMBINED RESULTS")
        print("=" * 50)
        print(f"Total files processed: {total_files}")
        print(f"QPU times collected: {len(all_qpu_times)}")
        print(f"Attack successes: {all_attack_success}")
        print(f"Attack success rate: {all_attack_success/total_files:.2%}")
        print(f"\nQPU Access Time Statistics:")
        print(f"  Mean: {mean_time:.6f} ms")
        print(f"  Standard Deviation: {std_time:.6f} ms")
        print(f"  Min: {min(all_qpu_times):.6f} ms")
        print(f"  Max: {max(all_qpu_times):.6f} ms")
        print(f"  Median: {np.median(all_qpu_times):.6f} ms")

        plt.figure(figsize=(15, 10))
        plt.subplot(2, 3, 3)

        n = len(all_qpu_times)
        x = np.arange(1, n + 1)
        y = np.asarray(all_qpu_times, dtype=float)

        plt.scatter(x, y, color='tab:blue', s=6, alpha=0.85,
                 linewidth=1.5, label='QPU Access Time')

        plt.fill_between(x, mean_time - std_time, mean_time + std_time,
                         alpha=0.2, color='orange')

        plt.axhline(mean_time, color='red', linestyle='--', linewidth=1.5,
                    alpha=0.85, label='Mean')

        k = 2.0
        y_min = max(0, mean_time - k * std_time)
        y_max = mean_time + k * std_time
        plt.ylim(y_min, y_max)
        plt.xlim(1, n)
        
        high_idxs = np.where(y > y_max)[0]
        if high_idxs.size > 0:
            marker_y_high = y_max - 0.7  # ▲ 위치

            for i, idx in enumerate(high_idxs):
                xx = idx + 1

                plt.plot(xx, marker_y_high, marker='^', color='crimson',
                         markersize=7, zorder=4)

                if i % 2 == 0:
                    label_y = y_max + 0.5
                    va = 'bottom'
                else:
                    label_y = y_max - 2.0
                    va = 'top'

                plt.text(xx, label_y, rf'\#{xx}',
                         color='crimson', fontsize=8,
                         ha='center', va=va, clip_on=False)

            highs_text = ", ".join([r"\#{}".format(i+1) for i in high_idxs])
            max_val = y[high_idxs].max()
            box = AnchoredText(
    f"High outliers : {highs_text}\nMax = {max_val:.2f} ms",
    loc='upper left',
    prop=dict(size=9, color='crimson'),
    frameon=True, pad=0.3, borderpad=0.4
)
            box.patch.set_alpha(0.85)
            box.patch.set_facecolor('white')
            plt.gca().add_artist(box)

        plt.xlabel('Test Number')
        plt.ylabel('QPU Access Time (ms)')
        plt.legend(loc='lower left')
        plt.grid(True, alpha=0.3)

        plt.tight_layout()
        plt.savefig('qpu_time_3.png', format='png', dpi=900, bbox_inches='tight')


if __name__ == "__main__":
    main()
```

```python
import matplotlib as mpl
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os

def main():

    img_paths = [
        "qpu_time_1.png",
        "qpu_time_2.png",
        "qpu_time_3.png",
    ]
    for p in img_paths:
        if not os.path.exists(p):
            raise FileNotFoundError(f"이미지 파일이 없습니다: {p}")

    imgs = [mpimg.imread(p) for p in img_paths]

    fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(18, 6))

    titles = [
    r"\fontsize{21}{24}\selectfont QPU Time Sequence -- \#KS=1",
    r"\fontsize{21}{24}\selectfont QPU Time Sequence -- \#KS=2",
    r"\fontsize{21}{24}\selectfont QPU Time Sequence -- \#KS=3",
]

    for ax, img, title in zip(axes, imgs, titles):
        ax.set_title(title, fontsize=21)
        ax.imshow(img)
        ax.set_title(title, fontsize=13)
        ax.axis("off")

    plt.tight_layout()

    out_pdf = "qpu_combined.pdf"
    out_png = "qpu_combined.png"  

    plt.savefig(out_pdf, format="pdf", dpi=900, bbox_inches="tight")
    plt.savefig(out_png, format="png", dpi=900, bbox_inches="tight")
    plt.show()

    print(f"Saved: {out_pdf}")
    print(f"Saved: {out_png}")

if __name__ == "__main__":
    main()
```
